<h1><strong>Seaborn</strong></h1>

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import pandas as pd
```

```python
tips = sns.load_dataset('tips')

tips
```

```python

fig,ax = plt.subplots()

sns.scatterplot(data=tips,x='total_bill',y='tip')
plt.show()
```

```python
fig,ax = plt.subplots(figsize=(9,6))

sns.scatterplot(
    data=tips,x='total_bill',y='tip',
    hue='time',style='smoker',size='size'
)
ax.set_title("Transaction Dynamics",fontsize=14,fontweight='bold')
ax.set_xlabel("Total Bill")
ax.set_ylabel("tip")

plt.show()
```

<h2>Line Plot</h2>

```python
fmri = sns.load_dataset('fmri')
fmri

fig,ax = plt.subplots()
sns.lineplot(data=fmri,x='timepoint',y='signal',ax=ax)
plt.show()
```

```python
fmri
```

```python
fig,ax = plt.subplots(figsize=(9,6))
sns.lineplot(data=fmri,x='timepoint',y='signal',hue='region',style='event',linewidth=2.5,palette='Set1',ax=ax)

ax.set_title("Neural Signal Tracking ")
ax.grid(True,linestyle='--',alpha=0.3)
plt.show()
```

<h2>Barplot</h2>

```python
fig,ax = plt.subplots()
sns.barplot(data=tips,x='day',y='total_bill',ax=ax)
plt.show()
```

```python
tips
```

```python
fig,ax = plt.subplots(figsize=(8,5))

sns.barplot(
    data=tips,
    x='day',y='total_bill',hue='sex',palette='Set2',
    edgecolor='0.2',ax=ax
)

ax.set_title("Average daily revenue")
ax.set_xlabel("Day of week")
ax.set_ylabel("Mean of total bill")
plt.show()
```

<h2>Count Plot</h2>

```python
fig,ax = plt.subplots()

sns.countplot(data=tips,x='day',ax=ax)
plt.show()
```

```python
fig,ax = plt.subplots(figsize=(8,5))

day_order = ['Thur','Fri','Sat','Sun']
sns.countplot(data=tips,x='day',order=day_order,palette='coolwarm',hue='sex',ax=ax)
ax.set_title("Customer traffic density per day")
plt.show()
```

<h2>Box plot</h2>

```python
fig,ax = plt.subplots()

sns.boxplot(data=tips,x='day',y='total_bill',ax=ax)

plt.show()
```

```python
fig,ax = plt.subplots(figsize=(9,5))

sns.boxplot(data=tips,x='day',y='total_bill',ax=ax,
hue='smoker',
palette='pastel',
flierprops={"marker":"x","markerfacecolor":"green"}
)

ax.set_title("Revenue variance and outlier profiling")
plt.show()
```

<h2>Violin plot</h2>

```python
fig,ax = plt.subplots()

sns.violinplot(data=tips,x='day',y='total_bill',ax=ax)

plt.show()
```

```python
fig,ax = plt.subplots(figsize=(9,5))

sns.violinplot(data=tips,x='day',y='total_bill',ax=ax,
hue='smoker',
palette='pastel'
)

ax.set_title("Revenue variance and outlier profiling")
plt.show()
```

<h2>Histogram</h2>

```python
fig,ax = plt.subplots()
sns.histplot(data=tips,x='total_bill',ax= ax )
plt.show()
```

```python
fig,ax = plt.subplots(figsize=(9,5))
sns.histplot(data=tips,x='total_bill',ax= ax,kde=True,hue='time',palette='magma',alpha=0.4 )
ax.set_title("Kernel daensity distribution")
plt.show()
```

<h2>Heatmap</h2>

```python
flights = sns.load_dataset('flights').pivot(index='month',columns='year',values='passengers')
flights
```

```python
fig,ax = plt.subplots()

sns.heatmap(data=flights,ax=ax)
plt.show()
```

```python
fig,ax = plt.subplots(figsize=(16,10))

sns.heatmap(data=flights,ax=ax,annot=True,cmap='YlGnBu',fmt='d',cbar_kws = {'label':"Passenger Count"},linewidths=0.5)
ax.set_title("Seasonal flight density over years")
plt.show()
```
