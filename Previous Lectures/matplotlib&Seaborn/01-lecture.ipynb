<h1>Matplotlib</h1>

<h2>Line Chart</h2>

Tracking changes or trends over continuous time interval or ordered categories

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,10,100)
y = np.sin(x)

fig,ax = plt.subplots()
ax.plot(x,y)
plt.show()
```

```python
fig,ax = plt.subplots(figsize=(10,5))
ax.plot(x,y,color='#1A73E8',linewidth=2.5,label='sine wave')
ax.plot(x,np.cos(x),color='#EA4335',linewidth=2,linestyle='--',label='cos wave')

ax.set_title("Trigonometric Functions",fontsize=16)
ax.set_xlabel("Time(s)",fontsize=12)
ax.set_ylabel("Amplitude",fontsize=12)
ax.legend()
ax.grid(True,linestyle=':')
plt.show()
```

<h2>Vertical Bar Chart</h2>



```python
categories = ['A','B','C','D','E']
values = [23,45,56,78,32]

fig,ax = plt.subplots()
ax.bar(categories,values)
plt.show()
```

```python
fig,ax = plt.subplots(figsize=(8,5))
bars = ax.bar(categories,values)
for bar in bars:
    yval = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2,yval + 2,f'{yval}',ha='center',va='bottom',fontsize=10)

ax.set_title('Category comparision')
plt.show()
```

```python
categories = ['A','B','C','D','E']
values = [23,45,56,78,32]

fig,ax = plt.subplots()
ax.barh(categories,values)
plt.show()
```

<h2>Scatter Plot</h2>

```python
x = np.random.randn(100)
y = np.random.randn(100)

fig,ax = plt.subplots()

ax.scatter(x,y)

plt.show()
```

```python
sizes = np.random.rand(100)*300
colors = np.random.rand(100)

fig,ax = plt.subplots(figsize=(9,6))
scatter = ax.scatter(x,y,c=colors,s=sizes,cmap='viridis',alpha=0.7,edgecolor='w')
ax.set_title("mulitvariate analysis")
cbar = fig.colorbar(scatter,ax=ax)
cbar.set_label("Intensity scale")
plt.show()
```

<h2>Histogram</h2>

```python
data = np.random.randn(1000)

fig,ax = plt.subplots()
ax.hist(data,bins=30,edgecolor='white')
plt.show()
```

<h2>Pie Chart</h2>

```python
sizes = [35,25,20,20]
labels = ['Python','SQL','Power BI','Tableau']

fig,ax = plt.subplots()
ax.pie(sizes,labels=labels)
plt.show()
```

```python
colors = ['#FF6B6B','#4D96FF','#6BCB77','#FFD93D']
explode = (0.05,0,0,0)

fig,ax = plt.subplots(figsize=(6,6))
ax.pie(sizes,explode=explode,labels=labels,colors = colors,autopct='%1.1f%%',shadow=True)
ax.set_title("Tech stack distribution")
plt.show()
```

```python
sizes = [35,25,20,20]
labels = ['Python','SQL','Power BI','Tableau']

fig,ax = plt.subplots()
ax.pie(sizes,labels=labels)
circle = plt.Circle((0,0),0.9,color='white')
ax.add_artist(circle)
plt.show()
```

```python
data = [np.random.normal(0,std,100) for std in range(1,4)]

fig,ax = plt.subplots()
ax.boxplot(data)
plt.show()
```

<h2>Area Chart</h2>

```python
x = np.arange(1,11)

y = [1,3,6,4,7,8,12,11,14,18]

fig,ax = plt.subplots()
ax.plot(x,y)
ax.fill_between(x,y,alpha=0.5)
plt.show()
```

```python
x = range(1,6)
y1 = [1,4,6,8,9]
y2 = [2,2,7,10,12]
y3 = [2,8,5,10,15]

fig,ax = plt.subplots()

ax.stackplot(x,y1,y2,y3,labels=['A','B','c'])
plt.legend(loc='upper left')
plt.show()
```

<h2>Grouped Bar Chart</h2>

```python
labels = ['G1','G2','G3','G4']
men_means = [20,34,30,35]
women_means = [25,32,34,20]

x = np.arange(len(labels))

width=0.35

fig,ax = plt.subplots()
ax.bar(x-width/2,men_means,width)
ax.bar(x+width/2,women_means,width)
plt.show()
```

```python
fig,ax = plt.subplots()
ax.bar(labels,men_means,width)
ax.bar(labels,women_means,width,bottom=men_means)
plt.show()
```

<h2>Step plot</h2>

```python
x = np.arange(7)
y = [1,3,2,5,4,6,5]

fig,ax = plt.subplots()
ax.step(x,y,where='mid',color='#D90429',linewidth=2.5,marker='s',markersize=6)
plt.show()
```

<h2>Error plot</h2>

```python
x = np.arange(1,5)
y = [10,15,12,18]
yerr = [1.2,2.5,0.8,1.9]

fig,ax = plt.subplots()
ax.errorbar(x,y,yerr=yerr)

plt.show()
```
