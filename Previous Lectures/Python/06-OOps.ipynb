<h3>Multiple Inheritence</h3>

```python
class Father:
    def skills(self):
        print("Drive")

class Mother:
    def skills(self):
        print("Cooking")

    def clean(self):
        print("clean")

class Child(Father,Mother):
    pass


c = Child()
c.clean()
```

<h3>Multilevel Inheritence</h3>

```python
class A:
    def methodA(self):
        print("A")

class B(A):
    def methodB(self):
        print("B")

class C(B):
    def methodC(self):
        print("C")

class D(C):
    def methodD(self):
        print("D")
```

```python
a = A()
b = B()
c = C()
d = D()


a.methodA()


b.methodB()
b.methodA()


c.methodC()
c.methodB()
c.methodA()

d.methodD()
d.methodC()
d.methodB()
d.methodA()
```

<h3>Method overriding(Runtime Polymorphism)</h3>

```python
class Animal:
    def sound(self):
        print("Animal makes sound.")

class Dog(Animal):
    def sound(self):
        print("Dog barks")

class Cat(Animal):
    def sound(self):
        print("Cat meows")


d = Dog()
d.sound()

Cat().sound()
```

```python
class Payment:
    def pay(self):
        print("Processing Payment")

    
class UPI(Payment):
    def pay(self):
        print("processing UPI Payment")

class Card(Payment):
    def pay(self):
        print("Processing Card Payment")

payments = [Payment(),UPI(),Card()]


for pay in payments:
    pay.pay()
```

<code>Question. </code>Shape -> Circle, Rectangle, Triangle, Square, area()

```python
class Shape:
    def area(self):
        print("Area of shape")

class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius
    def area(self):
        return 3.14 * self.radius ** 2
    
class Rectangle(Shape):
    def __init__(self,length,width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width
    
class Triangle(Shape):
    def __init__(self,base,height):
        self.base = base
        self.height = height

    def area(self):
        return 0.5 * self.base * self.height

class Square(Shape):
    def __init__(self,edge):
        self.edge = edge

    def area(self):
        return self.edge ** 2
    

areas = [Shape(),Circle(3),Rectangle(12,11),Triangle(3,4),Square(4)]

for area in areas:
    print(area.area())
```

<h1>5 Abstraction </h1>

```python
from abc import ABC,abstractmethod
class Shape:
    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def display(self):
        pass
```

```python
class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius
    def area(self):
        return 3.14 * self.radius ** 2
    
    def display(self):
        print("This is circle")
    
class Rectangle(Shape):
    def __init__(self,length,width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width
    
    def display(self):
        print("This is Rectangle")
    
class Triangle(Shape):
    def __init__(self,base,height):
        self.base = base
        self.height = height

    def area(self):
        return 0.5 * self.base * self.height
    
    def display(self):
        print("This is Triangle")

class Square(Shape):
    def __init__(self,edge):
        self.edge = edge

    def area(self):
        return self.edge ** 2
    
    def display(self):
        print("This is Square")
    
```

```python
areas = [Circle(3),Rectangle(12,11),Triangle(3,4),Square(4)]

for shape in areas:
    shape.display()
    print(shape.area())
```
