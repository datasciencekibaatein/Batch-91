<h2>Loops</h2>

while

```python
num = 1

while num <= 10:
    print(num)
    num = num  + 1
```

For loop

```python
for i in range(2,10,2):
    print(i)
```

```python
arr = [34,23,67,89,45,69]
```

```python
for num in arr:
    print(num)
```

```python
student = {
    "name":"Krish",
    "Age":24,
    45:"Value"
}
```

```python
student.keys()
```

```python
student.values()
```

```python
for key in student.keys():
    print(key)
```

```python
for value in student.values():
    print(value)
```

```python
student.items()
```

```python
for key,value in student.items():
    print(key,": ",value)
```

```python
students = {
    "Name":["Riddhi","Sohit","Rudra","Krish"],
    "Age":[34,23,12,34],
    "Course":["DS","BA","DA","AI"]

}
```

```python
for key,value in students.items():
    print(key,": ")
    for val in value:
        print(val)
```
