```python
import numpy as np
```

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(load_breast_cancer().data,load_breast_cancer().target,test_size=0.2,random_state=42)
```

```python
X_test
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(random_state=42)

model.fit(X_train,y_train)

y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import (accuracy_score,
                             confusion_matrix,
                             precision_score,
                             recall_score,
                             f1_score,
                             classification_report,
                             log_loss)
```

```python

print("Confusion Matrix\n",confusion_matrix(y_test,y_pred))
#Measures the proportion of correctly classified instances
print("accuracy score:\n ",accuracy_score(y_test,y_pred))
#Precision
#Measures how many predicted positive instances are actually positive
print("Precision:\n ",precision_score(y_test,y_pred))
#Recall
#Measures how many actual positive instances are correctly identified
print("Recall:\n ",recall_score(y_test,y_pred))
#F1-Score
#Harmonic mean of precision and recall
print("F1 Score \n ",f1_score(y_test,y_pred))
#Classification Report
print("Classification Report: \n",classification_report(y_test,y_pred))
```

```python
y_prob = model.predict_proba(X_test)
```
