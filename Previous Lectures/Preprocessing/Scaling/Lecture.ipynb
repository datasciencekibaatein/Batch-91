<h1>Feature Scaling</h1>

Feature scaling is the process of transforming numerical features so they are on the similar scale.
It helps many machine learning algorithms perform better and converge faster.

Algorithms based on distance or gradients may give more importance to salary.

Algorithms that need scaling:

- KNN
- K-Means
- SVM
- Logistic Regression
- Neural Networks
- PCA

Algorithms that usually don't nees scaling

- Decision Trees
- Random Forest
- XGBoost


```python
from sklearn.datasets import load_wine
import pandas as pd

wine = load_wine()

df = pd.DataFrame(
    wine.data,
    columns = wine.feature_names
)

df.head()
```

```python
df.describe().T[['min','max']]
```

1. <h2>Standard Scaling (z-score)</h2>



$$
Z = \frac{X - \mu}{\sigma}
$$

```python
df_sub = df[['proline','ash','malic_acid']]
```

```python
pl_mean = df_sub['proline'].mean()
pl_std = df_sub['proline'].std()

ash_mean = df_sub['ash'].mean()
ash_std = df_sub['ash'].std()

malic_acid_mean = df_sub['malic_acid'].mean()
malic_acid_std = df_sub['malic_acid'].std()
```

```python
df_sub.head()
```

```python
df_sub['proline_scaled'] = round((df_sub['proline'] - pl_mean)/pl_std,2)
```

```python
df_sub['ash_scaled'] = round((df_sub['ash']-ash_mean)/ash_std,2)
```

```python
df_sub['malic_acid_scaled'] = round((df_sub['malic_acid']-malic_acid_mean)/malic_acid_std,2)
```

```python
df_sub.describe().T[['min','max']]
```

Inverse transformation

```python
df_sub['proline_inverse'] = round(df_sub['proline_scaled'] * pl_std + pl_mean,1)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

scaled_data = scaler.fit_transform(df)

scaled_df = pd.DataFrame(scaled_data,columns=df.columns)

scaled_df.head()
```

```python
inverse_data = scaler.inverse_transform(scaled_df)

inverse_df = pd.DataFrame(inverse_data,columns=df.columns)

inverse_df.head()
```

2. <h2>Min Max Scaler</h2>

$$
X_{scaled} = \frac{X - X_{min}}{X_{max} - X_{min}}
$$

it ranges [0,1]

```python
df_ = df[['proline','ash','malic_acid']]

proline_min = df_['proline'].min()
proline_max = df_['proline'].max()

ash_min = df_['ash'].min()
ash_max = df_['ash'].max()

malic_acid_min = df_['malic_acid'].min()
malic_acid_max = df_['malic_acid'].max()
```

```python
df_['proline_scaled'] = round((df_['proline'] - proline_min)/(proline_max-proline_min),2)
```

```python
df_['ash_scaled'] = round((df_['ash']-ash_min)/(ash_max-ash_min),2)
```

```python
df_['malic_acid_scaled'] = round((df_['malic_acid']-malic_acid_min)/(malic_acid_max-malic_acid_min),2)
```

```python
df_.head()
```

```python
df_['proline_inverse']=df_['proline_scaled']*(proline_max-proline_min) + proline_min
```
