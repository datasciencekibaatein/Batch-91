We encounter cells that are empyt,null or undefined. These are called missing values - data points that were expected to record but were not recorded.

- Survey non-response
- System failure
- Data Entry error
- Deliberate omission
- Data migration

<h2>Types of missing values</h2>

- MCAR - Missing completely at random
- MAR - Missing at random
- MNAR - Missing not at random


<h3>MCAR</h3>
The probability of a value being missed has absolutely no relationship with any variable in the dataset.

<h3>MAR</h3>
The missingness is related to observed data, but not the missing values themselves

<h3>MNAR</h3>
The missingness depends on the observed value itself. This is hardest to handle.

<h2>Methods to handle missing values</h2>

1. Removing missing values(Deletion Technique)

    1.1 Row Deletion(Listwise deletion)
    
    1.2 Column Deletion(Dropping features)

2. Imputation

    2.1 Univariate Imputation
        
        - Numerical Value Imputation
        - Categorical value Imputation
    
    2.2 Multivariate Imputation

        - KNN Imputer
        - Iterative Imputer

```python
import pandas as pd
import numpy as np

data = {
    'Name':["Rohan","Krish","Bob","Sohit","Neha","Mayank"],
    'Age':[24,np.nan,34,24,np.nan,35],
    'Score':[85,90,np.nan,88,92,96],
    'Attendance':[100,90,80,np.nan,87,92]
}

df = pd.DataFrame(data)
df
```

```python
df.isnull().sum()
```

```python
round(df.isnull().mean()*100,2)
```

<h3>Row deletion</h3>

When it is good to use?

- Low percentage: if <5%  data is missing
- MCAR: The missingness has no hidden pattern
- Large Dataset: If you have million's of rows, losing 1000 rows won't hurt model's performance.


When not to use it?

- Small dataset: If you only have 100 rows, every single rows is precious.
- Bias Risk: If only 'High income' individuals refused  to answer the survey, dropping those rows removes your "High income" Representation entirely.

```python
df_dropped = df.dropna()
```

```python
df
```

```python
df_dropped
```

```python
df_subset = df.dropna(subset=['Score','Attendance'])
```
