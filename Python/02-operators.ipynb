Opertors

Operators in python are symbols , which helps us to perform any sort of operations(arithimatic , logical, assignment, bitwise etc.)

1. Arithimatic Operations

Addition

```python
num1 = 34
num2 = 42
print(num1+num2)
```

Subtraction

```python
print(num1-num2)
```

Multiplication

```python
print(num1 * num2)
```

Division

```python
print(num1/num2)
```

floor division

```python
print(num1//num2)
```

Power

```python
res = num1**2
print(res)
```

Modulus(remainder)

```python
res = 23%3
print(res)
```
