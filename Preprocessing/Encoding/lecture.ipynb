Categorical variables are variables that contain labels instead of numerical values.

### Types of categorical Variables

- Nominal Data
- Ordinal Data

```python
import seaborn as sns
import pandas as pd

df = sns.load_dataset("titanic")
df.head()
```

```python
df.isnull().sum()
```

```python
df['embark_town'].value_counts()
```

```python
df['deck'] = df['deck'].fillna("C")
```

```python
df['age'] = df['age'].fillna(df['age'].mean())
```

```python
df['embarked'] = df['embarked'].fillna("S")
```

```python
df['embark_town'] = df['embark_town'].fillna('Southampton')
```

```python
df.head(10)
```

### 1. Label Encoding

Assigns a unique integer to each category

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df['sex_encoded'] = le.fit_transform(df['sex'])

df[['sex','sex_encoded']].head()
```

```python
education = pd.DataFrame(
    {"education":["High School","Intermediate","Graduate","Post Graduate","PhD"]}
)

education['education_labelEncode'] = le.fit_transform(education['education'])

education
```

```python
from sklearn.preprocessing import OrdinalEncoder

categories = [["High School","Intermediate","Graduate","Post Graduate","PhD"]]

oe = OrdinalEncoder(categories=categories)

education[['ordinal_encoded']] = oe.fit_transform(education[['education']])

education
```

### 2. Ordinal Encoding

Used when categories have order.



```python
df['class'].value_counts()
```

```python
categories = [['First','Second','Third']]
```

```python
oe = OrdinalEncoder(categories=categories)

df[['class_encoded']] = oe.fit_transform(df[['class']])

df[['class','class_encoded']].head(10)
```

### 3. OneHotEncoder

Create separate column for category

```python
education = pd.DataFrame(
    {"education":["High School","Intermediate","Graduate","Post Graduate","PhD"]}
)
```

```python
education
```

```python
from sklearn.preprocessing import OneHotEncoder

ohe = OneHotEncoder(sparse_output=False)

encoded_data = ohe.fit_transform(education[['education']])

education_encoded = pd.DataFrame(encoded_data,columns=['Graduate','High School','Intermediate','PhD','Post Graduate'])
```

```python
education_encoded['education'] = education['education']
```

```python
education_encoded
```
