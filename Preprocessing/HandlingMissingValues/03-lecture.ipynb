### KNN Imputer

```python
import numpy as np
import pandas as pd

from sklearn.datasets import load_iris
from sklearn.impute import KNNImputer
```

```python
df = pd.DataFrame(load_iris().data,columns=load_iris().feature_names)
```

```python
df.isnull().mean()
```

```python
df.shape[1]
```

```python
df.loc[34,'sepal length (cm)']
```

```python
np.random.seed(42)

rows = np.random.randint(0,df.shape[0],30)

df.loc[rows,'sepal length (cm)'] = np.nan

rows = np.random.randint(0,df.shape[0],23)
df.loc[rows,'sepal width (cm)'] = np.nan

rows = np.random.randint(0,df.shape[0],34)
df.loc[rows,'petal length (cm)'] = np.nan

rows = np.random.randint(0,df.shape[0],12)
df.loc[rows,'petal width (cm)'] = np.nan
```

```python
round(df.isnull().mean()*100,2)
```

```python
imputer = KNNImputer(
    n_neighbors = 4,
    weights = 'distance'
)

filled = imputer.fit_transform(df)

filled_df = pd.DataFrame(filled,columns = df.columns)
```

```python
filled_df.isnull().mean()*100
```

### Iterative Imputer

```python
df = pd.DataFrame(load_iris().data,columns=load_iris().feature_names)
```

```python
np.random.seed(42)

rows = np.random.randint(0,df.shape[0],30)

df.loc[rows,'sepal length (cm)'] = np.nan

rows = np.random.randint(0,df.shape[0],23)
df.loc[rows,'sepal width (cm)'] = np.nan

rows = np.random.randint(0,df.shape[0],34)
df.loc[rows,'petal length (cm)'] = np.nan

rows = np.random.randint(0,df.shape[0],12)
df.loc[rows,'petal width (cm)'] = np.nan
```

```python
df.isnull().mean()*100
```

```python
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
```

```python
import warnings

warnings.filterwarnings('ignore')
```

```python
imputer = IterativeImputer(random_state=23)

filled = imputer.fit_transform(df)
```

```python
df_filled = pd.DataFrame(filled,columns = df.columns)

df_filled.isnull().mean()*100
```
