## Univariate Imputation

```python
import pandas as pd
import numpy as np

from sklearn.impute import SimpleImputer
```

```python


data = {
    'Name':["Rohan","Krish","Bob","Sohit","Neha","Mayank"],
    'Age':[24,np.nan,34,24,np.nan,35],
    'Score':[85,90,np.nan,88,92,96],
    'Attendance':[100,90,80,np.nan,87,92]
}

df = pd.DataFrame(data)
df
```

## Mean imputer

```python
mean_imputer = SimpleImputer(strategy='mean')

df_mean_imputed = mean_imputer.fit_transform(df[['Age','Score']])
```

```python
print(df_mean_imputed)
```

```python
df['Age'].mean()
```

## Median Imputer

```python
median_imputer = SimpleImputer(strategy='median')

df_meadian_imputed = median_imputer.fit_transform(df[['Age','Score']])
```

```python
pd.DataFrame(df_meadian_imputed,columns=['Age','Score'])
```

## most_frequent Imputer

```python
most_frequent_imputed = SimpleImputer(strategy='most_frequent')

df_most_frequent_imputed = most_frequent_imputed.fit_transform(df[['Age','Attendance']])
```

```python
df_most_frequent_imputed
```

```python
df['Notes'] = [np.nan,'Good',np.nan,'Average',np.nan,'Average']
```

```python
most_frequent_imputer_txt = SimpleImputer(strategy='most_frequent')

most_frequent_imputed_text = most_frequent_imputer_txt.fit_transform(df[['Score','Notes']])
```

```python
most_frequent_imputed_text
```

## Constant Imputer

```python
constant_imputer = SimpleImputer(strategy='constant',fill_value = -1)

df_constant_imputed = constant_imputer.fit_transform(df[['Score','Attendance']])
```

```python
df_constant_imputed
```

```python
constant_imputer_text = SimpleImputer(strategy='constant',fill_value="Not found")

df_constant_imputed_text = constant_imputer_text.fit_transform(df[['Notes']])
```

```python
df_constant_imputed_text
```
